```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
from sklearn.metrics import confusion_matrix,classification_report
from sklearn.metrics import  mean_squared_error
from sklearn.linear_model import LinearRegression
```

```python
df1=pd.read_csv("/content/Salary_Data.csv")
df1.head()
x=df1[["YearsExperience"]]
y=df1["Salary"]

x_train=x.iloc[:12]
x_test=x.iloc[12:]

y_train=y.iloc[:12]
y_test=y.iloc[12:]

model=LinearRegression()
model.fit(x_train,y_train)

print("Slope:",model.coef_[0])
print("Intercept",model.intercept_)
```

```python
df=pd.read_csv("/content/drive/MyDrive/Colab Notebooks/linear_regression_lab_15_records.csv")
df.head()
x=df[["Hours_Studied"]]
y=df["Marks"]

x_train=x.iloc[:12]
x_test=x.iloc[12:]

y_train=y.iloc[:12]
y_test=y.iloc[12:]

model=LinearRegression()
model.fit(x_train,y_train)

print("Slope:",model.coef_[0])
print("Intercept",model.intercept_)
```

```python
y_train_pred=model.predict(x_train)
y_test_pred=model.predict(x_test)

mse_train=mean_squared_error(y_train,y_train_pred)
mse_test=mean_squared_error(y_test,y_test_pred)

print("MSE Train:",mse_train)
print("MSE Test:",mse_test)

plt.scatter(x_train,y_train,color="blue",label="Training Data")
plt.scatter(x_test,y_test,color="green",label="Testing Data")
plt.plot(x_train,y_train_pred,color="red",label="Train Regression Line")
plt.plot(x_test,y_test_pred,color="black",label="Test Regression Line")
plt.xlabel("Hours_Studied")
plt.ylabel("Marks")
plt.legend()
plt.show()
```
